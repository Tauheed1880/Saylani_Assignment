# 1️⃣ What Problem RAG Actually Solves

### Core Limitation of LLMs

* Parametric memory only (weights-based knowledge)
* Hallucination under uncertainty
* Static knowledge cutoff
* No enterprise-private context

### RAG Solves:

* Knowledge grounding
* Real-time information retrieval
* Domain-specific augmentation
* Traceable responses (source citations)

Define clearly:

> RAG = LLM + External Knowledge Retrieval Layer

Not fine-tuning. Not prompt engineering.
It’s architectural augmentation.

---

# 2️⃣ High-Level Architecture of RAG

Basic Pipeline:

```
User Query
   ↓
Embedding Model
   ↓
Vector Search (Top-K retrieval)
   ↓
Context Injection
   ↓
LLM Generation
   ↓
Grounded Response
```

Break this into components:

### (A) Embedding Model

Transforms text → dense vectors
Examples:

* OpenAI text-embedding models
* Hugging Face sentence-transformers

Explain:

* Cosine similarity
* Vector space
* Dimensionality (e.g., 768, 1536)

---

### (B) Vector Database

Used for similarity search.

Examples:

* Pinecone
* Weaviate
* Chroma
* Milvus

Explain:

* Approximate Nearest Neighbor (ANN)
* HNSW indexing
* Top-K retrieval

---

### (C) Generator (LLM)

Examples:

* OpenAI models (GPT-4 class)
* Mistral AI models
* Meta LLaMA models

Clarify:

LLM does NOT search.
It only reasons over injected context.

# 3️⃣ Types of RAG Systems (Very Important Concept)

### 1. Naive RAG

Single query → retrieve → generate

### 2. Advanced RAG

* Query rewriting
* Hybrid search (BM25 + vector)
* Reranking
* Metadata filtering

### 3. Agentic RAG

* Multi-step reasoning
* Tool usage
* Iterative retrieval
* Self-reflection loop

This is where it connects to Agentic AI.

# 4️⃣ Explain Chunking (Most Students Ignore This)

Poor chunking = Bad RAG

Teach:

* Fixed-size chunking (e.g., 500 tokens)
* Recursive chunking
* Semantic chunking
* Overlap strategy (10–20%)

Explain why:

Too small → context fragmentation
Too large → retrieval noise

````python
import re
from typing import Literal
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

def extract_and_clean_pdf(pdf_path: str, mode: Literal['single', 'page'] = 'page'):
    loader = PyPDFLoader(pdf_path, mode=mode)
    docs = loader.load()

    # Clean the text inside each Document object directly
    for doc in docs:
        text = doc.page_content

        text = re.sub(r'(?i)page\s+\d+(\s+of\s+\d+)?', '', text)
        text = re.sub(r'[ \t]+', ' ', text)
        text = re.sub(r'(\w+)-\s*\n(\w+)', r'\1\2', text)
        text = re.sub(r'\n{3,}', '\n\n', text)

        doc.page_content = text.strip()

    # Return the list of Document objects
    return [d for d in docs if d.page_content]

# --- EXECUTION ---
pdf_file = "your_document.pdf"

# 1. This now returns a list of Document objects
cleaned_docs = extract_and_clean_pdf(pdf_file, mode = 'page')

# 2. Setup your text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

# 3. This works perfectly now because cleaned_docs is a list of Document objects
chunks = text_splitter.split_documents(cleaned_docs)

print(f"Created {len(chunks)} chunks")
````
